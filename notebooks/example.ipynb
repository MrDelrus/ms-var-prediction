```python
from ms_var_prediction import PROJECT_ROOT, MarkovSwitchingVaR, GaussianMixtureVaR, batch_backtest_model, fetch_valid_sp500_tickers

import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

```python
start_date = "2010-01-01"
end_date = "2021-01-01"
alpha = 0.5
window_shape = 250
output_dir = PROJECT_ROOT / "outputs"
```

# Markov Switching Model backtesting

Let's fetch valid for the selected time range tickers from current S&P 500:

```python
tickers = fetch_valid_sp500_tickers(start_date, end_date)

tickers_tiny = tickers[:10]
```

Now let's run backtest on them.

We can start with our baseline model - `GaussianMixture`:

```python
gmm_model = GaussianMixtureVaR()

gmm_res = batch_backtest_model(gmm_model, tickers_tiny, start_date, end_date, alpha, window_shape)
gmm_res.to_csv(output_dir / f"backtesting_gmm_{start_date}_{end_date}")
```

And now we can move on to our model - `MarkovSwitching`:

```python
msm_model = MarkovSwitchingVaR()

msm_res = batch_backtest_model(msm_model, tickers_tiny, start_date, end_date, alpha, window_shape)
msm_res.to_csv(output_dir / f"backtesting_msm_{start_date}_{end_date}")
```
